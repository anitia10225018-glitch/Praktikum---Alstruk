```python

# Menghitung Kembalian Belanja

# Input total belanja dan uang pembayaran
total_belanja = int(input("Masukkan total belanja: Rp"))
uang_bayar = int(input("Masukkan uang pembayaran: Rp"))

# Menghitung kembalian
if uang_bayar >= total_belanja:
    kembalian = uang_bayar - total_belanja
    print("\n=== STRUK BELANJA ===")
    print(f"Total belanja: Rp{total_belanja:,}")
    print(f"Uang pembayaran: Rp{uang_bayar:,}")
    print(f"Kembalian: Rp{kembalian:,}")
else:
    kekurangan = total_belanja - uang_bayar
    print("\nUang pembayaran tidak cukup.")
    print(f"Kekurangan pembayaran: Rp{kekurangan:,}")
```
