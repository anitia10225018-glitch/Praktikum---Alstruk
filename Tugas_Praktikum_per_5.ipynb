```python
# Praktikum 1 - Mengenali Tipe Data

nama = "Fajar"
umur = 19
berat = 54.5
terdaftar = True

# Menampilkan nilai dan tipe data
print("Nama:", nama)
print(type(nama))

print("Umur:", umur)
print(type(umur))

print("Berat:", berat)
print(type(berat))

print("Terdaftar:", terdaftar)
print(type(terdaftar))

# Mengubah nilai umur menjadi 20
umur = 20

print("\nSetelah umur diubah menjadi 20")

print("Nama:", nama)
print(type(nama))

print("Umur:", umur)
print(type(umur))

print("Berat:", berat)
print(type(berat))

print("Terdaftar:", terdaftar)
print(type(terdaftar))
```

```python
# Praktikum 2 - Operator Aritmatika

# Menentukan harga dan jumlah buku
harga = 12500
jumlah = 4

# Menghitung total harga
total = harga * jumlah

# Menghitung sisa pembagian
sisa = total % 2000

# Menghitung harga setelah diskon
diskon = total - 5000

# Menampilkan hasil perhitungan
print("Harga per buku: Rp", harga)
print("Jumlah buku:", jumlah)
print("Total harga: Rp", total)
print("Sisa pembagian total harga dengan 2000: Rp", sisa)
print("Harga setelah diskon Rp5000: Rp", diskon)
```

```python
# Praktikum 3 - Operator Perbandingan dan Logika

# Menentukan nilai dan kehadiran
nilai = 82
kehadiran = 90

# Memeriksa masing-masing syarat
print("Nilai memenuhi syarat:", nilai >= 75)
print("Kehadiran memenuhi syarat:", kehadiran >= 80)

# Memeriksa kedua syarat menggunakan and
print("Lulus dengan and:", nilai >= 75 and kehadiran >= 80)

# Mengubah nilai menjadi 70
nilai = 70

print("\nSetelah nilai diubah menjadi 70")
print("Nilai memenuhi syarat:", nilai >= 75)
print("Kehadiran memenuhi syarat:", kehadiran >= 80)
print("Lulus dengan and:", nilai >= 75 and kehadiran >= 80)

# Menggunakan operator or
print("Lulus dengan or:", nilai >= 75 or kehadiran >= 80)
```

```python
# Praktikum 4 - Operator Penugasan

# Menentukan poin awal
poin = 100
print("Nilai awal:", poin)

# Menambahkan 25 poin
poin += 25
print("Setelah ditambah 25:", poin)

# Mengurangi 10 poin
poin -= 10
print("Setelah dikurangi 10:", poin)

# Mengalikan poin dengan 2
poin *= 2
print("Setelah dikali 2:", poin)

# Membagi poin dengan 2
poin /= 2
print("Setelah dibagi 2:", poin)
```

```python
# Praktikum 5 - Struktur Data Dasar

# Membuat struktur data
nilai = [78, 85, 90, 85]
mahasiswa = {"nama": "Nadia", "kelas": "IF-A"}
koordinat = (5, 8)
angka = {2, 4, 4, 6, 8}

# Menampilkan nilai pertama dari list
print("Nilai pertama:", nilai[0])

# Menambahkan nilai 95 ke list
nilai.append(95)
print("List terbaru:", nilai)

# Menampilkan nama mahasiswa
print("Nama mahasiswa:", mahasiswa["nama"])

# Menampilkan koordinat pertama
print("Koordinat pertama:", koordinat[0])

# Menampilkan angka unik
print("Angka unik:", sorted(angka))
```

```python
# Tantangan Terintegrasi - Program Rekap Nilai

# Data mahasiswa pertama
nama = "Anitia Rahma"
nilai_tugas = 80
nilai_ujian = 90

rata_rata = (nilai_tugas + nilai_ujian) / 2
lulus = rata_rata >= 75

print("Nama mahasiswa:", nama)
print("Nilai tugas:", nilai_tugas)
print("Nilai ujian:", nilai_ujian)
print("Rata-rata:", rata_rata)
print("Status lulus:", lulus)

print("--------------------")

# Data mahasiswa kedua
nama = "Anitia Rahma"
nilai_tugas = 60
nilai_ujian = 70

rata_rata = (nilai_tugas + nilai_ujian) / 2
lulus = rata_rata >= 75

print("Nama mahasiswa:", nama)
print("Nilai tugas:", nilai_tugas)
print("Nilai ujian:", nilai_ujian)
print("Rata-rata:", rata_rata)
print("Status lulus:", lulus)
```
