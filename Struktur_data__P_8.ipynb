```python
#Array 1 dimensi

import array

#Membuat data suhu harian

suhu = array.array('i', [28, 30, 27, 32, 29])

print("Suhu harian:", suhu.tolist())

rata_rata = sum(suhu) / len(suhu)
print("Rata-rata suhu:", rata_rata)

print("Suhu tertinggi:", max(suhu))
print("Suhu terendah:", min(suhu))
```

```python
#Array 2 dimensi

import numpy as np

# Data nilai 3 mahasiswa untuk 3 mata kuliah
nilai = np.array([
    [80, 85, 90],
    [75, 80, 85],
    [90, 95, 88]
])

# Menampilkan data
print("Data Nilai Mahasiswa:")
print(nilai)

# Menghitung rata-rata setiap mahasiswa
for i in range(len(nilai)):
    rata_rata = np.mean(nilai[i])
    print("Rata-rata Mahasiswa", i + 1, ":", rata_rata)

# Menghitung rata-rata seluruh nilai
print("Rata-rata seluruh nilai:", np.mean(nilai))

# Mengambil nilai mahasiswa pertama pada mata kuliah pertama
print("Nilai mahasiswa pertama:", nilai[0][0])
```

```python
#Daftar antrian mahasiswa (Linked List)

class Node:
    def __init__(self, nama):
        self.nama = nama
        self.next = None

# Membuat node mahasiswa
m1 = Node("Anitia")
m2 = Node("Rizal")
m3 = Node("Devi")
m4 = Node("Alda")

# Menghubungkan node
m1.next = m2
m2.next = m3
m3.next = m4

# Menampilkan daftar antrean
print("=== DAFTAR ANTREAN MAHASISWA ===")

current = m1

while current:
    print(current.nama)
    current = current.next
```

```python

# Membuat tuple jadwal kuliah

jadwal = ("Senin", "Algoritma", "08:00", "B201 Lantai 2")

print("=== JADWAL KULIAH ===")
print("Hari:", jadwal[0])
print("Mata Kuliah:", jadwal[1])
print("Jam:", jadwal[2])
print("Ruangan:", jadwal[3])

# Menghitung jumlah data
print("Jumlah Data:", len(jadwal))

# Mengambil sebagian data
print("Informasi Kuliah:", jadwal[0:2])
```

```python
#PROGRAM: SISTEM PENGELOLAAN DATA MAHASISWA
#MENGGUNAKAN ARRAY, LINKED LIST, DAN TUPLE

import array

# ==================================
# 1. ARRAY
# Menyimpan nilai mahasiswa
# ==================================

nilai = array.array('i', [80, 85, 90, 75])

print("=== DATA NILAI MAHASISWA ===")
print("Daftar Nilai:", nilai.tolist())
print("Total Nilai:", sum(nilai))
print("Rata-rata:", sum(nilai) / len(nilai))
print("Nilai Tertinggi:", max(nilai))


# ==================================
# 2. LINKED LIST
# Menyimpan nama mahasiswa
# ==================================

class Node:
    def __init__(self, nama):
        self.nama = nama
        self.next = None

# Membuat node
m1 = Node("Anitia")
m2 = Node("Rizal")
m3 = Node("Devi")
m4 = Node("Alda")

# Menghubungkan node
m1.next = m2
m2.next = m3
m3.next = m4

print("\n=== DAFTAR MAHASISWA ===")

current = m1

while current:
    print(current.nama)
    current = current.next


# ==================================
# 3. TUPLE
# Menyimpan jadwal kuliah
# ==================================

jadwal = ("Senin", "Algoritma", "08:00", "B201 Lantai 2")

print("\n=== JADWAL KULIAH ===")
print("Hari:", jadwal[0])
print("Mata Kuliah:", jadwal[1])
print("Jam:", jadwal[2])
print("Ruangan:", jadwal[3])
print("Jumlah Data:", len(jadwal))
```
